# **1. Entendimento do Negócio (Business Understanding)**<a name="1"></a>

## 1.1. Pergunta de Negócio que queremos responder
O mercado de trabalho brasileiro passa por transformações estruturais contínuas. O objetivo central deste projeto é analisar a série temporal dos dados da PNAD Contínua (IBGE) para responder à seguinte pergunta central:

> A taxa de ocupação informal tem crescido a um ritmo superior ao da ocupação formal?

A hipótese a ser investigada é a de que as transformações e flexibilizações no mercado de trabalho têm impulsionado o crescimento da informalidade a taxas mais aceleradas em comparação com o emprego formalizado.

## 1.2. Público e uso pretendido da resposta
A resposta a essa pergunta e a entrega dos dados geram inteligência estratégica para os seguintes públicos:
- Órgãos Estatais, Governos Estaduais e Prefeituras: Uso para formulação, readequação e direcionamento de políticas públicas de geração de emprego, incentivo à formalização (como programas voltados ao MEI), fiscalização trabalhista e programas de assistência social.
- Empresas de Crédito e Instituições Financeiras: Uso na calibração de modelos de risco de crédito, bancarização e oferta de produtos financeiros adequados ao perfil de renda e estabilidade do trabalhador informal.
- Associações Comerciais e Setor Privado: Uso no dimensionamento do poder de compra e da dinâmica de consumo regional e nacional.

## 1.3. Forma da entrega final
A entrega deste projeto será consolidada na forma de um Dashboard Interativo, permitindo aos tomadores de decisão explorar a evolução temporal das taxas de ocupação, comparar formalidade versus informalidade e analisar tendências ao longo do tempo.

## 1.4. Critérios de sucesso do projeto
O projeto será considerado bem-sucedido se cumprir os seguintes critérios:
1. Validação de Hipótese embasada em Dados: Responder de forma conclusiva, utilizando dados estatisticamente válidos da PNAD Contínua, se a taxa de crescimento do trabalho informal superou ou não a do trabalho formal no período analisado.
2. Visualização Clara e Acessível: Entregar o dashboard funcional com indicadores de desempenho (KPIs) bem definidos, gráficos intuitivos de tendência e navegabilidade fluida.

# 2. **Entendimento dos Dados (Data Understanding)**<a name="2"></a>

## 2.1. Agregados e variáveis do IBGE explorados
Os dados são provenientes do IBGE (PNAD Contínua / Sistema SIDRA). Para responder ao problema do projeto, foram explorados os seguintes agregados e variáveis oficiais:

| Código | Variável / Indicador Oficial | Papel e significado no projeto |
| :--- | :--- | :--- |
| 12466 | Taxa de informalidade das pessoas de 14 anos ou mais de idade ocupadas na semana de referência | Métrica Chave: Percentual da população ocupada em situação de informalidade. |
| 4723 | Pessoas de 14 anos ou mais de idade ocupadas, em situação de informalidade, na semana de referência | Métrica Chave: Volume absoluto (em mil pessoas) de trabalhadores informais. |
| 4090 | Pessoas de 14 anos ou mais de idade ocupadas na semana de referência | Métrica Base: Total absoluto de trabalhadores no mercado (formais + informais). |
| 4099 | Taxa de desocupação, na semana de referência, das pessoas de 14 anos ou mais de idade | Contexto: Proporção de desempregados na força de trabalho. |
| 4088 | Pessoas de 14 anos ou mais de idade, na força de trabalho, na semana de referência | Contexto: Volume total de pessoas aptas a trabalhar (ocupadas + desocupadas). |
| 4092 | Pessoas de 14 anos ou mais de idade, desocupadas na semana de referência | Contexto: Volume absoluto de pessoas procurando emprego. |
| 4094 | Pessoas de 14 anos ou mais de idade, fora da força de trabalho, na semana de referência | Contexto: Volume de pessoas em idade de trabalhar que não estão no mercado. |
| 4096 | Taxa de participação na força de trabalho, na semana de referência | Análise: Percentual da população em idade ativa que compõe a força de trabalho. |
| 4097 | Nível da ocupação, na semana de referência, das pessoas de 14 anos ou mais | Análise: Proporção da população em idade ativa efetivamente ocupada. |
| 4098 | Nível da desocupação, na semana de referência, das pessoas de 14 anos ou mais | Análise: Razão entre a população desocupada e a população em idade ativa. |

## 2.2. Método de coleta
Os dados são coletados por técnicos do IBGE por meio de visitas presenciais ou chamadas telefônicas, sendo inseridos diretamente em computadores de mão — DMC (Dispositivo Móvel de Coleta) — com questionários eletrônicos. No projeto, a extração dos dados consolidados é realizada via API do SIDRA/IBGE.

## 2.3. Granularidade disponível
- Granularidade Geográfica: Nível N2, referente à esfera nacional (Brasil).
- Granularidade Temporal: Dados Trimestrais, cobrindo o período do 1º trimestre de 2019 ao 2º trimestre de 2026.

## 2.4. Problemas de qualidade identificados
Ao explorar o retorno bruto da API do IBGE/SIDRA, foram mapeados os seguintes problemas de qualidade:
- Valores Ausentes / Defasagem de Divulgação: Risco de trimestres mais recentes ainda não estarem publicados no momento da requisição, retornando valores nulos (NaN/null).
- Formatação de Atributos e Tipos Inadequados: Os valores numéricos e percentuais são retornados pela API formatados como texto (string), exigindo conversão explícita.
- Metadados Redundantes e Inconsistências de Layout: A API entrega os dados em um formato pivotado contendo cabeçalhos complexos, códigos de classificação genéricos e notas de rodapé, exigindo limpeza estrutural para conversão em formato tabular limpo (tidy data).

# 3. **Preparação dos Dados (Data Preparation)**<a name="3"></a>

## 3.1. Seleção dos dados
Os dados já estão pré-selecionados com base nos identificadores oficiais do IBGE para a PNAD Contínua, cobrindo a série temporal nacional de 2019 a 2026.

## 3.2. Limpeza e tratamentos necessários
Os dados faltantes e eventuais inconsistências na série temporal serão tratados diretamente nas células de código do notebook do projeto. Serão aplicadas conversões de tipos (texto para float/int) e eliminação de metadados desnecessários do JSON do IBGE.

## 3.3. Construção de novas variáveis
Serão construídas as variáveis `pop_formal` (População Ocupada Formal) e `taxa_formalidade_pct` (Taxa de Formalidade) para permitir uma avaliação mais aprofundada dos dados. A população formal é calculada pela diferença entre a população ocupada total (código 4090) e a população informal (código 4723), viabilizando a comparação direta da taxa de crescimento de ambos os setores.

## 3.4. Integração dos dados
Temos apenas uma fonte de dados (a API do sistema SIDRA/IBGE para a PNAD Contínua), portanto não é necessário realizar processos de integração ou junção entre diferentes bases de dados.

## 3.5. Formato final dos dados (Esquema de tabela e dicionário de Dados)
Alguns dados serão tratados para um formato mais útil. O conjunto de dados final assume o seguinte esquema tabular e dicionário de colunas:

| Nome da Coluna | Tipo de Dado | Origem / Regra de Formatação | Descrição |
| :--- | :--- | :--- | :--- |
| id_trimestre | Integer | Formato AAAATQ (Ex: 201901) | Chave primária para ordenação temporal contínua. |
| data_inicio_trimestre | Date | Padrão ISO (YYYY-MM-DD) | Data do primeiro dia do trimestre para eixos de gráficos. |
| ano | Integer | Extraído da API IBGE | Ano de referência do dado. |
| trimestre | Integer | Extraído da API IBGE | Número do trimestre (1 a 4). |
| pop_ocupada_total | Integer | Variável 4090 (IBGE) | Total de pessoas ocupadas (em milhares). |
| pop_informal | Integer | Variável 4723 (IBGE) | Total de pessoas ocupadas em informalidade (em milhares). |
| pop_formal | Integer | Calculado: pop_ocupada_total - pop_informal | Total de pessoas ocupadas formalmente (em milhares). |
| taxa_informalidade_pct | Float | Variável 12466 (IBGE) | Percentual da população ocupada na informalidade (%). |
| taxa_formalidade_pct | Float | Calculado: (pop_formal / pop_ocupada_total) * 100 | Percentual da população ocupada na formalidade (%). |
| taxa_desocupacao_pct | Float | Variável 4099 (IBGE) | Taxa de desocupação (%). |
| pop_forca_trabalho | Integer | Variável 4088 (IBGE) | Total de pessoas na força de trabalho (em milhares). |
| pop_desocupada | Integer | Variável 4092 (IBGE) | Total de pessoas desocupadas (em milhares). |
| pop_fora_forca | Integer | Variável 4094 (IBGE) | Total de pessoas fora da força de trabalho (em milhares). |

## 3.6. Critérios para dados "Prontos" (Data Quality Acceptance)
O grupo considerará a base de dados pronta para ser consumida pelo dashboard quando atender estritamente aos seguintes critérios:
1. Completude da Série Temporal: Ausência de lacunas de trimestres no intervalo de 1º trim/2019 ao 2º trim/2026.
2. Consistência Matemática: A equação pop_formal + pop_informal = pop_ocupada_total deve ser verdadeira para 100% dos registros.
3. Integridade de Tipagem: Ausência de valores nulos (NaN) em colunas obrigatórias e conversão total de campos numéricos/datas sem caracteres especiais.

## 3.7. Execução recorrente do pipeline e versionamento de Código
- Automação do Pipeline Recorrente: O pipeline de extração, tratamento e geração da tabela final será codificado em Python (requests e pandas). A atualização ocorrerá trimestralmente via rotina agendada (ex.: GitHub Actions ou Cron Job), que consultará a API do IBGE em busca de novos trimestres divulgados, reprocessará as regras de negócio e atualizará a base consumida pelo dashboard.
- Versionamento do Código: Todo o código-fonte (scripts de pré-processamento, consultas e documentação) será mantido em um repositório Git/GitHub. A organização adotará branches separadas para desenvolvimento (dev) e para a versão estável em produção (main), garantindo rastreabilidade e histórico de alterações na lógica de pré-processamento.
